## 1.1 Preprocessing Steps

The raw social-media posts were cleaned and normalized through the following preprocessing pipeline:

#### 1. Emoji Normalization
Convert emojis into readable tokens using `emoji.demojize()` so that emotion-related symbols (e.g., 😂 → `:face_with_tears_of_joy:`) are preserved as meaningful features.

#### 2. Text Normalization
Apply lowercasing and expand common contractions (e.g., *don't → do not*, *I'm → I am*) to reduce vocabulary sparsity and preserve negation cues.

#### 3. Noise Removal
Remove URLs, user mentions, and non-alphabetic symbols while keeping emoji tokens to eliminate noise irrelevant to emotion classification.

#### 4. Character Repetition Normalization
Normalize elongated words commonly found in social posts (e.g., *loooove → loove*) to prevent unnecessary vocabulary fragmentation.

#### 5. Stopword Removal with Negation Preservation
Remove standard English stopwords except negation words (*not, no, nor*), since negation directly influences emotional polarity.

#### 6. Tokenization and Reconstruction
Tokenize cleaned text, filter noise, and reconstruct normalized text into a consistent format suitable for TF-IDF vectorization.

#### 7. Dataset Loading and Label Merging
Load posts from JSON, flatten nested structures with `json_normalize`, split accounts into train/test using `data_identification.csv`, and merge emotion labels by post ID.


## 1.2 Feature Engineering Steps

After preprocessing the raw text, we transformed the cleaned sentences into numerical features suitable for machine learning models. The following steps summarize the feature engineering process:

#### 1. Train/Validation Split
The training set was further split into training and validation subsets using a 90/10 stratified split.  
This ensures that each emotion category is proportionally represented in both partitions.

#### 2. TF-IDF Vectorization
We used `TfidfVectorizer` from scikit-learn to convert text into high-dimensional sparse vectors.  
The following design choices were applied:

- **Word-level analysis** (`analyzer="word"`)
- **1–2 gram features** to capture short phrases relevant to emotion (e.g., *"not happy"*)
- **Stopword removal** using the built-in English list
- **min_df = 2** to remove extremely rare terms  
- **max_df = 0.95** to remove extremely common, uninformative terms  
- **sublinear_tf = True** for log-scaled term frequency  
- **max_features = 300,000** to limit feature space size  

The TF-IDF model was fit only on the training set to avoid information leakage:
- `X_tr_vec = tfidf.fit_transform(X_tr)`
- `X_val_vec = tfidf.transform(X_val)`
- `X_test_vec = tfidf.transform(X_test)`

#### 3. Final Output Representation
Each text sample is represented as a sparse TF-IDF vector containing unigram and bigram features.  
These vectors form the input to all downstream machine learning models used in this project.


### 1.3 Explanation of Your Model

To perform emotion classification, we adopted a linear Support Vector Machine (LinearSVC) trained on TF-IDF features.  
This model is widely used for text classification due to its efficiency in high-dimensional sparse spaces.

#### 1. Model Choice: LinearSVC
LinearSVC was selected because:

- It handles high-dimensional TF-IDF vectors effectively.
- It is computationally efficient compared to nonlinear SVMs.
- It works well for short-text classification tasks.
- The linear decision boundary aligns well with bag-of-words-based representations.

The model used the following configuration:
- **C = 2.0** (regularization strength)
- **Hinge loss formulation**
- **One-vs-rest multi-class strategy** (default for LinearSVC)

#### 2. Handling Class Imbalance
The emotion labels in the dataset are imbalanced.

## 1. Preprocessing Steps

````python
import pandas as pd
import numpy as np
import datasets as Dataset
import json
import re
import emoji
from collections import Counter
import nltk
from nltk.corpus import stopwords
STOPWORDS = set(stopwords.words("english"))
````

````python
def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    # Convert emojis to readable tokens
    text = emoji.demojize(text, delimiters=(" ", " "))
    
    # Lowercase
    text = text.lower()
    
    # Replace common contractions
    text = re.sub(r"n't", " not", text)
    text = re.sub(r"'re", " are", text)
    text = re.sub(r"'s", " is", text)
    text = re.sub(r"'ll", " will", text)
    text = re.sub(r"'m", " am", text)
    
    # Remove URLs
    text = re.sub(r"http\S+|www\.\S+", "", text)

    # Remove mentions
    text = re.sub(r"@\w+", "", text)
    
    # Keep only alphabetic words + emoji tokens + emotion tokens
    text = re.sub(r"[^a-zA-Z\s:]", " ", text)
    
    # Normalize repeated chars (loooove → loove)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # Remove stopwords EXCEPT negation words
    tokens = text.split()
    tokens = [
        t for t in tokens
        if t not in STOPWORDS or t in {"not", "no", "nor"}
    ]
    
    text = " ".join(tokens)
    
    return text
````

````python
with open("final_posts.json", "r", encoding="utf-8") as f:
    data = json.load(f)

df = pd.json_normalize(data, sep='.')


df = df.rename(
    columns={
        "root._type": "type",
        "root._source.post.post_id": "id",
        "root._source.post.text": "text",
        "root._source.post.hashtags": "hashtags"
    }
)
print(df.head())
````

### Split the accounts into train / test

````python
df1 = pd.read_csv("data_identification.csv")
print(df1.head())

df_train = df1[df1["split"] == "train"]
df_test = df1[df1["split"] == "test"]

print(df_train.shape)
print(df_test.shape)
````

### Load emotion label

````python
df_emotion = pd.read_csv("emotion.csv")
print(df_emotion.head())
````

### Merge emotion label

````python
df_train_full = df_train.merge(df_emotion, on="id", how="left")
df_train_full = df_train_full.merge(df[["id", "text"]], on="id", how="left")

df_test_full = df_test.merge(df[["id", "text"]], on="id", how="left")

print(df_train_full.head())
print("\n")
print(df_test_full.head())
````

````python
# Split X and y
X_train = df_train_full["text"].fillna("")
y_train = df_train_full["emotion"]

X_test = df_test_full["text"].fillna("")
````

## 2. Feature Engineering Steps

````python
### Add the code related to the feature engineering steps in cells inside this section
````

````python
# Split validation and training
from sklearn.model_selection import train_test_split
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train,
    test_size=0.1,  # 10% for validation
    stratify=y_train,
    random_state=42
)
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    analyzer="word",
    ngram_range=(1, 2),
    stop_words="english",
    min_df=2,
    max_df=0.95,
    sublinear_tf=True,
    lowercase=True,
    max_features=300000
)


X_tr_vec = tfidf.fit_transform(X_tr)
X_val_vec = tfidf.transform(X_val)
X_test_vec = tfidf.transform(X_test)
````

## 3. Model Implementation Steps

````python
### Add the code related to the model implementation steps in cells inside this section
````

````python
from sklearn.utils.class_weight import compute_class_weight
from sklearn.svm import LinearSVC


classes = np.unique(y_tr)
weights = compute_class_weight('balanced', classes=classes, y=y_tr)

class_weight_dict = dict(zip(classes, weights))

clf = LinearSVC(C=2, class_weight=class_weight_dict)

clf.fit(X_tr_vec, y_tr)
````

````python
from sklearn.metrics import classification_report

y_pred_val = clf.predict(X_val_vec)
print(classification_report(y_val, y_pred_val))
````
